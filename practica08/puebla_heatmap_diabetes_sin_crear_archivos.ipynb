## 4. Algoritmo de clasificación de riesgo de diabetes

El algoritmo asigna puntos según factores de riesgo:

- Edad mayor o igual a 45 años
- IMC elevado
- Presión arterial elevada
- Colesterol LDL alto
- Glucosa en ayunas elevada
- Actividad física baja
- Antecedente familiar de diabetes

Después convierte el puntaje en una categoría de riesgo.


```python
def clasificar_riesgo_diabetes(row):
    puntos = 0

    # Edad
    if row['edad'] >= 45:
        puntos += 1
    if row['edad'] >= 60:
        puntos += 1

    # IMC
    if row['imc'] >= 25:
        puntos += 1
    if row['imc'] >= 30:
        puntos += 2

    # Presión arterial
    if row['presion_sistolica'] >= 130 or row['presion_diastolica'] >= 80:
        puntos += 1
    if row['presion_sistolica'] >= 140 or row['presion_diastolica'] >= 90:
        puntos += 1

    # Colesterol LDL
    if row['colesterol_ldl'] >= 130:
        puntos += 1
    if row['colesterol_ldl'] >= 160:
        puntos += 1

    # Glucosa en ayunas
    if row['glucosa_ayunas'] >= 100:
        puntos += 2
    if row['glucosa_ayunas'] >= 126:
        puntos += 3

    # Actividad física
    if row['actividad_fisica'] == 'Baja':
        puntos += 1

    # Antecedente familiar
    if row['antecedente_familiar'] == 'Sí':
        puntos += 1

    if row['glucosa_ayunas'] >= 126:
        categoria = 'Diabetes probable'
    elif row['glucosa_ayunas'] >= 100:
        categoria = 'Prediabetes'
    elif puntos <= 2:
        categoria = 'Sin riesgo'
    elif puntos <= 4:
        categoria = 'Riesgo bajo'
    elif puntos <= 6:
        categoria = 'Riesgo de diabetes'
    else:
        categoria = 'Alto riesgo de diabetes'

    return pd.Series([puntos, categoria])

pacientes_df[['puntaje_riesgo', 'categoria_riesgo']] = pacientes_df.apply(clasificar_riesgo_diabetes, axis=1)
pacientes_df.head()
```

## 5. Asignar peso para el mapa de calor

Mientras más alto sea el riesgo, más peso tendrá el paciente en el mapa de calor.


```python
pesos_riesgo = {
    'Sin riesgo': 1,
    'Riesgo bajo': 2,
    'Riesgo de diabetes': 3,
    'Prediabetes': 4,
    'Alto riesgo de diabetes': 5,
    'Diabetes probable': 6
}

pacientes_df['peso_mapa'] = pacientes_df['categoria_riesgo'].map(pesos_riesgo)
pacientes_df[['paciente_id', 'municipio', 'categoria_riesgo', 'puntaje_riesgo', 'peso_mapa']].head()
```

## 6. Tabla general de pacientes por categoría de riesgo


```python
tabla_riesgo = (
    pacientes_df['categoria_riesgo']
    .value_counts()
    .rename_axis('Categoria')
    .reset_index(name='Total de pacientes')
)

tabla_riesgo['Porcentaje'] = (tabla_riesgo['Total de pacientes'] / len(pacientes_df) * 100).round(2)
tabla_riesgo
```

## 7. Tabla de pacientes por municipio y categoría


```python
tabla_municipio_riesgo = (
    pacientes_df
    .groupby(['municipio', 'categoria_riesgo'])
    .size()
    .reset_index(name='total_pacientes')
    .sort_values(['municipio', 'total_pacientes'], ascending=[True, False])
)

tabla_municipio_riesgo.head(20)
```

## 8. Municipios con mayor puntaje de riesgo


```python
riesgo_por_municipio = (
    pacientes_df
    .groupby('municipio')
    .agg(
        total_pacientes=('paciente_id', 'count'),
        promedio_puntaje=('puntaje_riesgo', 'mean'),
        peso_total=('peso_mapa', 'sum'),
        latitud=('Latitude', 'mean'),
        longitud=('Longitude', 'mean')
    )
    .reset_index()
    .sort_values('peso_total', ascending=False)
)

riesgo_por_municipio.head(15)
```

## 13. Mapa por municipio con colores según categoría dominante


```python
orden_riesgo = {
    'Sin riesgo': 1,
    'Riesgo bajo': 2,
    'Riesgo de diabetes': 3,
    'Prediabetes': 4,
    'Alto riesgo de diabetes': 5,
    'Diabetes probable': 6
}

colores = {
    'Sin riesgo': 'green',
    'Riesgo bajo': 'lightgreen',
    'Riesgo de diabetes': 'orange',
    'Prediabetes': 'red',
    'Alto riesgo de diabetes': 'darkred',
    'Diabetes probable': 'black'
}

categoria_dominante = (
    pacientes_df
    .groupby(['municipio', 'categoria_riesgo'])
    .size()
    .reset_index(name='total')
    .sort_values(['municipio', 'total'], ascending=[True, False])
    .drop_duplicates('municipio')
)

municipio_resumen = riesgo_por_municipio.merge(categoria_dominante, on='municipio', how='left')

mapa_categorias = folium.Map(location=centro_puebla, zoom_start=8, tiles='cartodbdark_matter')

for _, row in municipio_resumen.iterrows():
    categoria = row['categoria_riesgo']
    folium.CircleMarker(
        location=[row['latitud'], row['longitud']],
        radius=max(4, min(14, row['total_pacientes'] / 8)),
        color=colores.get(categoria, 'blue'),
        fill=True,
        fill_opacity=0.75,
        popup=(
            f"<b>Municipio:</b> {row['municipio']}<br>"
            f"<b>Categoría dominante:</b> {categoria}<br>"
            f"<b>Pacientes:</b> {row['total_pacientes']}<br>"
            f"<b>Puntaje promedio:</b> {row['promedio_puntaje']:.2f}"
        ),
        tooltip=f"{row['municipio']} - {categoria}"
    ).add_to(mapa_categorias)

# mapa_categorias.save('05_mapa_categorias_riesgo_puebla.html')
mapa_categorias
```
